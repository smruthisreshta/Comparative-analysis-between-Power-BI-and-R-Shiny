```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("sales_dataset.csv")

# --- 1. Revenue by Product ---
plt.figure(figsize=(8,5))
df.groupby("Product")["Revenue"].sum().sort_values().plot(kind="bar", color="skyblue")
plt.title("Total Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()

```

```python
# --- 2. Revenue Trend Over Time ---

df["Date"] = pd.to_datetime(df["Date"])
revenue_time = df.groupby("Date")["Revenue"].sum()

plt.figure(figsize=(10,5))
plt.plot(revenue_time.index, revenue_time.values, color="blue", linewidth=2)
plt.title("Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

```python
# --- 3. Revenue by Region ---
plt.figure(figsize=(6,6))
df.groupby("Region")["Revenue"].sum().plot(kind="pie", autopct="%1.1f%%", startangle=90, cmap="Set3")
plt.title("Revenue by Region")
plt.ylabel("")
plt.show()
```

```python
# --- 4. Salesperson Performance ---
plt.figure(figsize=(8,5))
df.groupby("Salesperson")["Revenue"].sum().sort_values().plot(kind="barh", color="green")
plt.title("Revenue by Salesperson")
plt.xlabel("Revenue")
plt.ylabel("Salesperson")
plt.show()
```
